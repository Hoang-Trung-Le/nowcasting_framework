```python
def load_colocation_data(metadata_file):
    with open(metadata_file, "r") as f:
        metadata = json.load(f)
    return metadata


def load_site_data(aqms_filepath=None, pas_filepaths=None):
    """
    Load AQMS and PAS data from the specified file paths.

    Args:
        aqms_filepath (str): Path to the AQMS data file.
        pas_filepaths (dict): Dictionary of PAS file paths with sensor IDs as keys.

    Returns:
        pd.DataFrame, dict: AQMS data and dictionary of PAS data.
    """
    # Load AQMS data
    aqms_data = pd.read_csv(aqms_filepath, index_col=0, parse_dates=True)

    # Load PAS data into a dictionary
    pas_data = {}
    if pas_filepaths:
        for sensor_id, filepath in pas_filepaths.items():
            pas_data[sensor_id] = pd.read_csv(filepath, index_col=0, parse_dates=True)

    return aqms_data, pas_data
```

```python
def calculate_annual_total(data, parameter):
    """
    Calculate the total annual value/concentration for each year.

    Args:
        data (pd.DataFrame): DataFrame with datetime index and parameter columns.
        parameter (str): The parameter to calculate the total annual value for.

    Returns:
        pd.Series: Annual total values for the parameter.
    """
    return data[parameter].resample("YE").sum()


def calculate_monthly_proportions(data, parameter):
    """
    Calculate the proportion of each monthly value with respect to the total annual amount.

    Args:
        data (pd.DataFrame): DataFrame with datetime index and parameter columns.
        parameter (str): The parameter to calculate the proportions for.

    Returns:
        pd.DataFrame: Proportions of each monthly value with respect to the total annual amount.
    """
    monthly_totals = data[parameter].resample("M").sum()
    annual_totals = calculate_annual_total(data, parameter)

    proportions = monthly_totals / annual_totals.reindex(
        monthly_totals.index, method="ffill"
    )
    return proportions


def calculate_average_proportions(proportions):
    """
    Calculate the average proportion of each monthly value with respect to all proportions of that same month.

    Args:
        proportions (pd.DataFrame): DataFrame of monthly proportions.

    Returns:
        pd.Series: Average proportions for each month.
    """
    return proportions.groupby(proportions.index.month).mean()


def calculate_baseline_values(data, parameter):
    """
    Calculate the baseline values for each month for each year.

    Args:
        data (pd.DataFrame): DataFrame with datetime index and parameter columns.
        parameter (str): The parameter to calculate the baseline values for.

    Returns:
        pd.DataFrame: Baseline values for each month for each year.
    """
    proportions = calculate_monthly_proportions(data, parameter)
    average_proportions = calculate_average_proportions(proportions)
    annual_totals = calculate_annual_total(data, parameter)

    print("average proportions:", average_proportions)
    print("annual total:", annual_totals)

    baseline = pd.DataFrame()
    for year, year_total in annual_totals.items():
        monthly_index = pd.date_range(start=f"{year.year}-01-01", periods=12, freq="M")
        baseline_year = average_proportions.values * year_total
        baseline = pd.concat([baseline, pd.Series(baseline_year, index=monthly_index)])

    return baseline


def create_baseline_model(data, parameter):
    """
    Create a baseline model for the AQMS data based on monthly values.

    Args:
        data (pd.DataFrame): AQMS data with a datetime index.
        parameter (str): The parameter for which to create the baseline model.

    Returns:
        pd.Series: The baseline model.
    """
    return calculate_baseline_values(data, parameter)


def load_and_prepare_all_sites(metadata_file, freq):
    metadata = load_colocation_data(metadata_file)

    for location, details in metadata.items():
        aqms_details = details.get("AQMS", {})
        site_id = details["AQMS"]["site_id"]
        aqms_start_date, aqms_end_date = aqms_details["data_range"]

        # sensors = details.get("PAS", {})
        # pas_filepaths = {}
        # for sensor_id, sensor_details in sensors.items():
        # sensor_start_date, sensor_end_date = sensor_details["data_range"]

        start_date = aqms_start_date
        end_date = aqms_end_date

        start_date_formatted = start_date.replace("-", "")
        end_date_formatted = end_date.replace("-", "")
        aqms_filepath = f"../../../data/processed/colocation/{location}/AQMS/{start_date_formatted}_{end_date_formatted}/aqms_{site_id}_{start_date_formatted}_{end_date_formatted}_{freq}.csv"
        # pas_filepath = f"../../../data/processed/colocation/{location}/PAS/{sensor_id}/{start_date_formatted}_{end_date_formatted}/pas_{sensor_id}_{start_date_formatted}_{end_date_formatted}_{freq}.csv"
        # pas_filepaths[sensor_id] = pas_filepath

        aqms_data, _ = load_site_data(aqms_filepath)
        print("Site: ", location)
        parameters = list(HEADER_MAPPING.keys())

        for parameter in parameters:
            baseline = create_baseline_model(aqms_data, parameter)

            print("baseline plot data: ", baseline)

            if location == "Wagga_Wagga_North":
                location = "Wagga_Wagga"
            folderpathdir = os.path.join(
                "fig",
                "climatology",
                "colocation",
                location,
                "baseline",
                f"{start_date_formatted}_{end_date_formatted}",
                parameter,
            )

            if not os.path.exists(folderpathdir):
                os.makedirs(folderpathdir)

            filename = os.path.join(
                folderpathdir,
                f"{location}_{start_date_formatted}_{end_date_formatted}_{freq}.png",
            )
            print("filename: ", filename)
            baseline.columns = ["baseline"]
            baseline_filtered = baseline[baseline.index.year < 2024]
            print(baseline)
            plot_time_series(
                baseline_filtered,
                y_columns=["baseline"],
                xlabel="Time",
                ylabel=f"{parameter}",
                title=f"{freq.capitalize()} baseline model for {parameter} at {location} station",
                date_format="%Y-%m",
                figsize=(16, 4),
                save_dir=filename,
            )


metadata_file = "../../../metadata/colocation/colocation.json"
freq = "monthly"
load_and_prepare_all_sites(metadata_file, freq)

# # Generate some dummy data for testing
# dates = pd.date_range(start="2010-01-01", end="2020-12-31", freq="M")
# np.random.seed(42)
# dummy_data = pd.DataFrame(
#     {"datetime_utc": dates, "PM2.5": np.random.randint(10, 100, len(dates))}
# )
# dummy_data.set_index("datetime_utc", inplace=True)
# print("dummy:", dummy_data)
# # Example usage
# parameter = "PM2.5"
# baseline = create_baseline_model(dummy_data, parameter)
# print(baseline)
```
